## 作業說明

本作業使用 Python 製作聚餐費用試算器，練習字串處理、資料型別轉換、數值計算及布林判斷。


```python

# HW08 第一組：原始資料

event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300
```

```python

# 整理聚餐名稱
clean_event_name = event_name.strip().lower()

print(clean_event_name)
```

```python

# 整理原始餐費
clean_meal_amount = meal_amount.strip()
clean_meal_amount = clean_meal_amount.replace("NT$", "")
clean_meal_amount = clean_meal_amount.replace(",", "")

print(clean_meal_amount)
```

```python

# 將文字轉換成數字
meal_amount_num = int(clean_meal_amount)
people_num = int(people)

print(meal_amount_num)
print(people_num)
```

```python

# 計算服務費
service_fee = meal_amount_num * service_rate

print(service_fee)
```

```python

# 計算折價後總金額
total_amount = meal_amount_num + service_fee - discount

print(total_amount)
```

```python

# 計算每人分攤金額
amount_per_person = total_amount / people_num

print(amount_per_person)
```

```python

# 判斷是否超過每人預算
over_budget = amount_per_person > budget_per_person

print(over_budget)
```

```python

# 第一組：完整費用摘要

print("===== 聚餐費用試算結果 =====")
print("聚餐名稱：", clean_event_name)
print("原始餐費：", meal_amount_num, "元")
print("服務費：", service_fee, "元")
print("折價券：", discount, "元")
print("應付總額：", total_amount, "元")
print("分攤人數：", people_num, "人")
print("每人分攤：", amount_per_person, "元")
print("每人預算：", budget_per_person, "元")
print("是否超過預算：", over_budget)
print("============================")
```
